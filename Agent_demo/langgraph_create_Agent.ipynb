```python
from langchain_core.tools import tool
from langgraph.prebuilt import ToolNode, create_react_agent

# 定义工具 return_direct=True 表示直接返回工具结果，不用大模型进行进一步处理
@tool("devide_tool", return_direct=True)
def devide(a: int, b: int) -> float:
    """
    计算两个证整数的除法
    Args：
        a(int): 除数
        b(int): 被除数
    """
    # 自定义错误
    if b == 1:
        raise ValueError("被除数不能为1")
    return a / b
print(devide.name)
print(devide.description)
print(devide.args)

# 定义工具错误处理函数
def handle_tool_error(error: Exception) -> str:
    """
    处理工具错误
    Args：
        error(Exception): 工具错误
    """
    if isinstance(error, ValueError):
        return "被除数为1没有意义， 请重新输入一个除数和被除数。"
    elif isinstance(error, ZeroDivisionError):
        return "被除数不能为0， 请重新输入一个除数和被除数。"
    return f"工具调用错误：{error}"

tool_node = ToolNode(
    [devide],
    handle_tool_errors=handle_tool_error
)
agent_with_error_handle = create_react_agent(  # 旧方法，create_react_agent不再维护
    model=llm,
    tools=tool_node
)
result = agent_with_error_handle.invoke({"messages": [{"role": "user", "content": "计算10除以1等于多少？"}]})
# 打印最后的返回结果
# print(result["messages"][-1].content)
print(result)
```

### 短期记忆CheckPoint

```python
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.prebuilt import create_react_agent

checkpointer = InMemorySaver()

def get_weather(city: str) -> str:
    """
    获取你某个城市的天气
    :param city:
    :return:
    """
    return f"城市：{city}，天气一直都是晴天！"

agent_checkpoint = create_react_agent(
    model=llm,
    tools=[get_weather],
    checkpointer=checkpointer
)
# Run the Agent
config = {
    "configurable":{
        "thread_id": "1"
    }
}

cs_response = agent_checkpoint.invoke(
    {
        "messages": [{"role": "user", "content": "长沙天气怎么样？"}]
    },
    config
)
print(cs_response)
print("-----------------")
# Continue the conversation using the same thread_id
bj_response = agent_checkpoint.invoke(
    {
        "messages": [{"role": "user", "content": "北京呢？"}]
    },
    config
)
print(bj_response)
```

```python
# 一、Summarization 总结
from langmem.short_term import SummarizationNode
from langchain_core.messages.utils import count_tokens_approximately
from langgraph.prebuilt import create_react_agent
from langchain.agents import AgentState
from langgraph.checkpoint.memory import InMemorySaver
from typing import Any

# 使用大模型对历史信息进行总计
summarization_node = SummarizationNode(
    token_counter=count_tokens_approximately,
    model=llm,
    max_tokens=384,
    max_summary_tokens=128,
    output_messages_key="llm_input_messages",
)

class State(AgentState):
    # 注意：这个状态管理的作用是为了能够保存上一次总结的结果。这样可以防止每次调用大模型时，都要重新总结历史信息
    # 这是一个比较常见的优化方式，因为大模型的调用是比较耗时的
    context: dict[str, Any]

checkpointer = InMemorySaver()

agent = create_react_agent(
    model=llm,
    tools=tools,
    pre_model_hook=summarization_node, # 此函数将在每次调用LLM的节点之前被调用
    state_schema=State,
    checkpointer=checkpointer,
)


# 二、另外，还提供了trim_messages函数，用于定期清理压缩短期记忆。
from langchain_core.messages.utils import trim_messages, count_tokens_approximately
from langgraph.prebuilt import create_react_agent
# 此函数将清理压缩历史消息
def trim_messages_hook(state):
    trimmed_messages = trim_messages(
        state["messages"],
        strategy="last",
        token_counter=count_tokens_approximately,
        max_tokens=384,
        start_on="human",
        end_on=("human", "tool"),
    )
    return {"llm_input_messages": trimmed_messages}

checkpointer = InMemorySaver()
agent = create_react_agent(
    model=llm,
    tools=[],
    pre_model_hook=trim_messages_hook, # 此函数将在每次调用LLM的节点之前被调用
    checkpointer=checkpointer,
)
```

```python
from typing import Annotated
from langgraph.prebuilt import InjectedState
from langchain.agents import create_agent
from langchain.agents import AgentState
from langchain_core.tools import tool

class CustomState(AgentState):
    user_id: str

@tool(return_direct=True)
def get_user_info(
        state: Annotated[CustomState, InjectedState],
) -> str:
    """查询用户信息"""
    user_id = state["user_id"]
    return f"{user_id}用户的姓名：张三！" if user_id == "user_123" else "未知用户"

agent = create_agent(
    model=llm,
    tools=[get_user_info],
    state_schema=CustomState
)
agent.invoke(
    {
        "messages": "查询用户信息",
        "user_id": "user_123"
    }
)
```

### 长期记忆Store

```python
from langchain_core.runnables import RunnableConfig
from langgraph.config import get_store
from langchain.agents import create_agent
from langgraph.store.memory import InMemoryStore
from langchain_core.tools import tool

# 定义长期存储
store = InMemoryStore()
# 添加一些测试数据。users是命名空间，user_123是key，后面的JSON数据是value
store.put(
    ("users",), # namespace 名称空间
    "user_123", # key 键
    {
        "name": "张三",
        "age": 39,
    } # value 值
)

# 定义工具
@tool(return_direct=True)
def get_user_info(config: RunnableConfig) -> str:
    """查询用户信息"""
    # 获取长期存储，获取到了之后，这个存储组件可读可写
    store = get_store()
    #     store.put(
    #     ("users",), # namespace 名称空间
    #     "user_123", # key 键
    #     {
    #         "name": "张三",
    #         "age": 39,
    #     } # value 值
    # )
    # 获取配置中的用户ID
    user_id = config["configurable"].get("user_id")
    user_info = store.get(("users",), user_id)
    return str(user_info.value) if user_info else "Unknown user"

agent = create_agent(
    model=llm,
    tools=[get_user_info],
    store=store, # 指定长期存储组件
)

agent.invoke(
    {
        "messages":[
            {
                "role": "user",
                "content": "帮我查找用户信息",
            }
        ]
    },
    config={"configurable":{"user_id":"user_123"}}
)
```

### Human-in-the-loop人类监督（人机协同/人在回路）

```python
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.types import interrupt
from langchain.agents import create_agent
from langchain_core.tools import tool

# An example of a sensitive tool that requires human review /approval 一个需要人工审查/批准的敏感工具示例
@tool(return_direct=True)
def book_hotel(hotel_name: str) -> str:
    """预订酒店"""
    response = interrupt(f"正准备执行'book_hotel'工具预定酒店，相关参数名：{{'hotel_name': {hotel_name}}}."
                         f"请选择OK，表示同意，或者edit，提出补充意见。"
                         )
    if response["type"] == "OK":
        pass
    elif response["type"] == "edit":
        hotel_name = response["args"]["hotel_name"]
    else:
        raise ValueError(f"Unknown response type: {response['type']}")
    return f"成功在{hotel_name}预定了一个房间。"

checkpointer = InMemorySaver()

agent = create_agent(
    model=llm,
    tools=[book_hotel],
    checkpointer=checkpointer,
)

config = {
    "configurable": {
        "thread_id": "user_123", # thread_id 要保持相同
    }
}

for chunk in agent.stream(
    {"messages":[{"role": "user", "content": "帮我在图灵宾馆预订一个房间"}]},
    config,
):
    print(chunk)
    print("="*50)
```

执行完成后，会在book_hotel执行过程中，输出一个Interrupt响应，表示当前正在等待用户输入确认。
接下来，可以通过Agent提交一个Command请求，来继续完成之前的任务。
需要注意的是，在这个示例中，Agent只会一直等待用户输入。如果等待时间过长，后续请求就无法恢复了。

```python
from langgraph.types import Command
for chunk in agent.stream(
    # Command(resume={"type": "OK"}),
    Command(resume={"type": "edit", "args": {"hotel_name": "三号宾馆"}}),
    config
):
    print(chunk)
    print(chunk["tools"]["messages"][-1].content)
    print("&"*50)
```
